# Feature Engineering

Separate Features and Target

```python
x = df.drop('class', axis=1)
y = df['class']
```

```python
print("Features (X) shape:", x.shape)
print("Target (y) shape:", y.shape)
```

Check Feature Data Types

```python
#Check Feature Data Types
print("Feature Data Types:")
print(x.dtypes)
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x)
```

```python
print(X_scaled)
```

```python
print(X_scaled.shape)
```

Split the Data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled,
                                                    y
                                                    ,test_size=0.2,
                                                      random_state=42)
```

Verify Shapes

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

# Model Building

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)

```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
```

Random forest  

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train )
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
```

SVM

```python
from sklearn.svm import SVC

svm =SVC()
svm.fit(X_train, y_train)
```

# Model Evaluation
Make Predictions

Logistic Regression:

```python
y_pred_lr = lr.predict(X_test)
```

Decision Tree:

```python
y_pred_dt = dt.predict(X_test)
```

Random Forest:

```python
y_pred_rf = rf.predict(X_test)
```

KNN:

```python
y_pred_knn = knn.predict(X_test)
```

SVM

```python
y_pred_SVM = svm.predict(X_test)
```

Calculate Accuracy

```python
from sklearn.metrics import accuracy_score

print("LR:", accuracy_score(y_test, y_pred_lr)) 
print("DT:", accuracy_score(y_test, y_pred_dt)) 
print("RF:", accuracy_score(y_test, y_pred_rf)) 
print("KNN:", accuracy_score(y_test, y_pred_knn)) 
print("SVM:", accuracy_score(y_test, y_pred_SVM)) 
```

Classification Report (Best Model)

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_rf))
```

Confusion Matrix (Best Model)

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_rf)
print(cm)
```

Confusion Matrix Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score

comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'Decision Tree', 'Random Forest', 'KNN', 'SVM'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_knn),
        accuracy_score(y_test, y_pred_SVM)
    ]
})

comparison = comparison.sort_values(by='Accuracy', ascending=False)

comparison
```
